```python
# ==========================================
# Task 1: Credit Scoring Model
# ==========================================
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from sklearn.datasets import make_classification
import warnings
warnings.filterwarnings('ignore')

print("--- Step 1: Generating/Loading Financial Classification Dataset ---")
# 1. Generating synthetic financial history data for credit scoring
X, y = make_classification(n_samples=1000, n_features=8, n_classes=2, random_state=42)

# Converting to DataFrame to look like professional financial/credit data
column_names = ['Income', 'ExistingDebts', 'PaymentHistory', 'CreditUtilization', 'EmploymentLength', 'Age', 'NumCreditCards', 'LoanAmount']
df = pd.DataFrame(X, columns=column_names)
df['Creditworthiness'] = y # Target variable (0: Low Risk/Good, 1: High Risk/Bad)

print(df.head())

print("\n--- Step 2: Data Splitting ---")
# 2. Splitting into Training (80%) and Testing (20%) Sets
X_train, X_test, y_train, y_test = train_test_split(df.drop('Creditworthiness', axis=1), df['Creditworthiness'], test_size=0.2, random_state=42)

print(f"Training Set Shape: {X_train.shape}")
print(f"Testing Set Shape: {X_test.shape}")

print("\n--- Step 3: Model Training (Random Forest Classifier) ---")
# 3. Training Random Forest Model for Credit Risk Prediction
model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(X_train, y_train)
print("Model training completed successfully!")

print("\n--- Step 4: Model Evaluation ---")
# 4. Evaluating Predictions
y_pred = model.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)

print(f"Credit Scoring Model Accuracy: {accuracy * 100:.2f}%\n")
print("Classification Report:")
print(classification_report(y_test, y_pred))

print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred))
```
